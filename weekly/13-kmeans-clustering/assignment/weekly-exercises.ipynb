## Exercise 3


The plot below shows data generated from **five** clusters that have been stretched and rotated. Suppose you did
not know how many clusters there were.

```python
from sklearn.datasets import make_blobs

# Five spherical clusters, then stretched and rotated into elliptical ones
X_blobs, y_blobs = make_blobs(n_samples=300, centers=5, cluster_std=2.0, random_state=42)
transformation_matrix = np.array([[0.6, -0.6], [-0.2, 0.9]])
X_aniso = np.dot(X_blobs, transformation_matrix)

plt.figure(figsize=(8, 8))
plt.scatter(X_aniso[:, 0], X_aniso[:, 1], c=y_blobs, cmap='viridis', marker='o', edgecolors='k', s=60)
plt.title('Ground truth')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.show()
```
