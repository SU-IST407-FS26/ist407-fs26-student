## Exercise 2


Explore what PCA keeps and what it throws away, using the synthetic data below. There are three features. In `x`
and `y` the two classes are identical (wide noise); they differ only in `z`, which has a much smaller spread.
The cell below builds the data, splits it, and reports KNN's test accuracy on all three features.

```python
np.random.seed(0)
n_points = 500

# Class 0
x0 = np.random.normal(loc=0, scale=10, size=n_points)
y0 = np.random.normal(loc=0, scale=10, size=n_points)
z0 = np.random.normal(loc=0, scale=1, size=n_points)

# Class 1
x1 = np.random.normal(loc=0, scale=10, size=n_points)
y1 = np.random.normal(loc=0, scale=10, size=n_points)
z1 = np.random.normal(loc=2, scale=1, size=n_points)

X_syn = np.vstack((np.hstack((x0, x1)), np.hstack((y0, y1)), np.hstack((z0, z1)))).T
y_syn = np.hstack((np.zeros(n_points), np.ones(n_points)))

X_train, X_test, y_train, y_test = train_test_split(X_syn, y_syn, test_size=0.2, random_state=42)

knn = KNeighborsClassifier().fit(X_train, y_train)
print(f"All three features: {accuracy_score(y_test, knn.predict(X_test)):.3f}")
```
